```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['label']=encoder.fit_transform(df.label)
```

```python
from sklearn.model_selection  import train_test_split 

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train_sc = sc.fit_transform(x_train)
X_test_sc = sc.transform(x_test)
```

```python
X_train_sc.shape
```

```python
import keras
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import Dropout
classifier = Sequential()
```

```python
classifier.add(Dense(units=64,
                     activation='relu',
                     input_shape=(X_train_sc.shape[1],)))
```

```python
classifier.add(Dropout(rate=0.1))
```

```python



classifier.add(Dense(units=64,
                     activation='relu'))

classifier.add(Dropout(rate=0.1))
```

```python
classifier.add(Dense(units=22, activation='softmax'))
```

```python
classifier.compile(optimizer="Adam", loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
classifier.fit(X_train_sc, y_train, batch_size=32, epochs=200)
```

```python
import numpy as np
def recomand(a,b,c,d,e,f,g):
    input_data=np.array([[a,b,c,d,e,f,g]])
    scaled_input=sc.transform(input_data)
    prediction=classifier.predict(scaled_input,verbose=True)
    predicted_class_idx=np.argmax(prediction)
    recomand_crop_name= encoder.inverse_transform([predicted_class_idx])
    confidence= prediction[0][predicted_class_idx]*100
    return recomand_crop_name[0],confidence
    
    
```

```python
import gradio as gr
import numpy as np

emoji = {
    "rice": "🌾", "maize": "🌽", "chickpea": "🫘", "kidneybeans": "🫘", "pigeonpeas": "🫘",
    "mothbeans": "🫘", "mungbean": "🫘", "blackgram": "🫘", "lentil": "🫘",
    "pomegranate": "🍎", "banana": "🍌", "mango": "🥭", "grapes": "🍇", "watermelon": "🍉",
    "muskmelon": "🍈", "apple": "🍎", "orange": "🍊", "papaya": "🥭", "coconut": "🥥",
    "cotton": "☁️", "jute": "🌿", "coffee": "☕"
}

def predict(n, p, k, temp, hum, ph, rain):
    data = sc.transform(np.array([[n, p, k, temp, hum, ph, rain]]))
    probs = classifier.predict(data, verbose=0)[0]
    names = encoder.classes_
    top = int(np.argmax(probs))
    crop = names[top]
    headline = f"## {emoji.get(crop, '🌱')} Recommended crop: **{crop.title()}**\nConfidence: **{probs[top] * 100:.1f}%**"
    scores = {names[i]: float(probs[i]) for i in range(len(names))}
    return headline, scores

with gr.Blocks(theme=gr.themes.Soft(primary_hue="green", secondary_hue="emerald"), title="Crop Recommendation") as demo:
    gr.Markdown(
        "# 🌱 Smart Crop Recommendation System\n"
        "An Artificial Neural Network (Keras + PyTorch) that suggests the best crop "
        "from soil nutrients and weather conditions."
    )
    with gr.Row():
        with gr.Column():
            gr.Markdown("### 🧪 Soil nutrients")
            n = gr.Slider(0, 140, value=90, label="Nitrogen (N)")
            p = gr.Slider(5, 145, value=42, label="Phosphorus (P)")

            k = gr.Slider(5, 205, value=43, label="Potassium (K)")
            gr.Markdown("### 🌦️ Climate")
            temp = gr.Slider(8, 44, value=21, label="Temperature (°C)")
            hum = gr.Slider(14, 100, value=82, label="Humidity (%)")
            ph = gr.Slider(3.5, 9.9, value=6.5, label="Soil pH")
            rain = gr.Slider(20, 300, value=203, label="Rainfall (mm)")
            btn = gr.Button("🔍 Recommend Crop", variant="primary")
        with gr.Column():
            result = gr.Markdown("## Your recommendation will appear here")
            chart = gr.Label(num_top_classes=3, label="Top 3 predictions")

    inputs = [n, p, k, temp, hum, ph, rain]
    btn.click(predict, inputs=inputs, outputs=[result, chart])

    gr.Examples(
        examples=[
            [90, 42, 43, 21, 82, 6.5, 203],
            [20, 130, 200, 23, 83, 6.0, 70],
            [100, 18, 30, 25, 55, 6.7, 95],
            [118, 30, 50, 27, 80, 6.9, 175],
        ],
        inputs=inputs,
    )

demo.launch()
```
